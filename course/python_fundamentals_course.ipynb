## 1. Variables & types

A **variable** is a name that stores a value. You create one with `=` (assignment).

> 💡 **Excel equivalent**: naming a cell or a range, then reusing that name in formulas.

````python
account = "FR7610011000601234567890185"   # text  -> str
amount = 1540.75                          # decimal number -> float
count = 12                                # whole number   -> int
is_flagged = True                         # yes/no -> bool

print(account)
print("Amount:", amount, "| Count:", count, "| Flagged:", is_flagged)
````

Python knows the **type** of each value. The four you will meet most often:

| Type | Meaning | Example |
|---|---|---|
| `str` | text (string) | `"EUR"` |
| `int` | whole number | `12` |
| `float` | decimal number | `1540.75` |
| `bool` | true / false | `True`, `False` |

Use `type()` to check, and the type name to **convert**.

````python
print(type(amount))          # <class 'float'>
print(type(count))           # <class 'int'>

# Convert between types
print(int(amount))           # 1540  (drops the decimals)
print(float(count))          # 12.0
print(str(count))            # "12"  (text, useful to build labels)
print("Amount is " + str(amount) + " EUR")   # text + text
````

## 2. Strings & f-strings

**Strings** are text, written between quotes (`"..."` or `'...'`).
The cleanest way to insert values into text is an **f-string**: a string prefixed with `f`,
where `{...}` is replaced by the value inside.

> 💡 **Excel equivalent**: the `&` operator or `CONCAT()` / `TEXT()` for formatting.

````python
currency = "EUR"
amount = 9950.5

# f-string: put an f before the quotes, values go in { }
print(f"Transaction of {amount} {currency}")

# Format numbers inside the f-string:
print(f"Amount: {amount:,.2f} {currency}")   # 9,950.50  -> thousands sep + 2 decimals
print(f"Flagged: {amount > 9000}")            # you can even put an expression
````

Strings come with handy **methods** (functions attached with a dot). A few auditors use a lot:

````python
label = "  Virement International  "

print(label.strip())            # remove surrounding spaces -> "Virement International"
print(label.strip().upper())    # to UPPERCASE
print(label.strip().lower())    # to lowercase
print("Virement".replace("Virement", "Transfer"))   # find & replace
print("FR7610011000".startswith("FR"))               # True
print("ALPHA SARL".split(" "))  # cut into a list on spaces -> ['ALPHA', 'SARL']
print(len("EUR"))               # number of characters -> 3
````

## 3. Lists

A **list** is an ordered collection of values, written between **square brackets** `[ ]`,
separated by commas. It is the everyday container in Python: column names, country codes,
extraction results…

> 💡 **Excel equivalent**: a range of values in a single column, or an enumeration in a formula.

````python
countries = ['France', 'Luxembourg', 'Panama', 'Cyprus', 'Malta', 'Singapore']
print(countries)
print("Number of items:", len(countries))
````

````python
# Access by position (the index starts at 0)
print(countries[0])    # first item
print(countries[-1])   # last item
print(countries[1:4])  # slice: index 1 up to (but not including) 4

# Membership test
print('Panama' in countries)     # True
print('Germany' in countries)    # False
print(countries.index('Cyprus')) # position of 'Cyprus'
````

````python
work = countries.copy()          # work on a copy so the original is untouched

work.append('Germany')           # add ONE item at the end
work.extend(['Belgium', 'Italy'])# add SEVERAL items
work.remove('Malta')             # remove the first matching value
last = work.pop()                # remove AND return the last item
print("Removed:", last)
print("Result:", work)

print(sorted(countries))         # a new sorted list (original unchanged)
````

### List comprehensions

A **comprehension** builds or filters a list in one elegant line — one of Python's most useful tools.

```
[expression   for item in iterable   if condition]
```

````python
countries = ['France', 'Luxembourg', 'Panama', 'Cyprus', 'Malta', 'Singapore']

upper = [c.upper() for c in countries]              # transform each item
long_names = [c for c in countries if len(c) > 6]   # filter with a condition
codes = [c[:2].upper() for c in countries]          # transform + slice

print(upper)
print(long_names)
print(codes)

# Audit case: keep amounts above a threshold
amounts = [1200.0, 9800.0, 450.5, 7500.0, 9950.0, 150.0]
large = [a for a in amounts if a > 5000]
print("Amounts > 5000:", large)
````

## 4. Dictionaries

A **dictionary** stores **key → value** pairs, written between **curly braces** `{ }`.
You look values up **by key** (not by position). Perfect for lookup tables, mappings,
and describing one record.

> 💡 **Excel equivalent**: a two-column lookup table used by `VLOOKUP` — the key column
> and the value column.

````python
# Exchange rates to EUR (key = currency, value = rate)
rates = {"EUR": 1.0, "USD": 0.92, "GBP": 1.17}

print(rates["USD"])              # look up by key -> 0.92
print(rates.get("CHF", "n/a"))   # .get avoids an error if the key is missing

rates["CHF"] = 1.05              # add / update a pair
print(rates)
print("Currencies:", list(rates.keys()))     # all keys
print("Rates:", list(rates.values()))        # all values
````

````python
# A dictionary can describe one record (like one row)
transaction = {
    "id": 358,
    "amount": 9950.0,
    "currency": "USD",
    "country": "Panama",
}
print(transaction["country"])
print(f"Tx {transaction['id']}: {transaction['amount']} {transaction['currency']}")

# This is exactly how you build a pandas DataFrame:
#   pd.DataFrame({"amount": [...], "currency": [...]})  -> keys become column names
````

## 5. Tuples

A **tuple** is like a list but **immutable** (it cannot be changed after creation).
Written with parentheses `( )`. Used for fixed groups of values — e.g. coordinates,
or a `(rows, columns)` shape.

> 💡 In pandas, `df.shape` returns a tuple like `(506, 10)`.

````python
shape = (506, 10)          # a tuple: (rows, columns)
print("Rows:", shape[0])   # access by position, like a list
print("Columns:", shape[1])

# Unpacking: assign each element to its own variable in one line
rows, columns = shape
print(f"{rows} rows and {columns} columns")

# A tuple is read-only:  shape[0] = 999  would raise an error
````

## 6. Booleans, comparisons & logical operators

A **boolean** is either `True` or `False`. Comparisons produce booleans, and you combine
them with **`and` / `or` / `not`**. This is the backbone of every filter you will write in pandas.

> 💡 **Excel equivalent**: `=A1>1000`, and the `AND()`, `OR()`, `NOT()` functions.

````python
amount = 9950
country = "Panama"

print(amount > 9000)               # True
print(amount == 10000)             # ==  means "is equal to"   -> False
print(amount != 10000)             # !=  means "is different"  -> True

# Combine conditions
print(amount > 9000 and amount < 10000)     # both must be true
print(country == "Panama" or country == "Cyprus")
print(not (amount > 10000))                  # invert
````

**`None`** is a special value meaning "nothing / no value" (the ancestor of pandas' `NaN`
for missing data). Test it with `is None`.

````python
counterparty = None            # value not provided
print(counterparty is None)    # True
print(counterparty is not None)# False
````

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | different from |
| `>` `<` | greater / less than |
| `>=` `<=` | greater-or-equal / less-or-equal |
| `and` | both conditions true |
| `or` | at least one true |
| `not` | invert |
| `in` | is a member of |

## 7. Conditions (`if` / `elif` / `else`)

Run different code depending on a condition. Note the **colon `:`** and the **indentation**
(4 spaces) that defines the block.

> 💡 **Excel equivalent**: the `IF()` function, nested for several cases.

````python
amount = 9950

if amount >= 10000:
    category = "To declare"
elif amount >= 9000:                 # elif = "else if" (checked only if the first was false)
    category = "Just below threshold"
else:
    category = "Standard"

print(category)
````

> ⚠️ Indentation is **not decoration** in Python: it defines what is inside the block.
> A misaligned line changes the meaning or raises an error.

## 8. Loops (`for`, `while`)

A **loop** repeats code for each item of a collection. The `for` loop is the most common.

> 💡 **Excel equivalent**: copying a formula down a whole column — the same operation
> repeated for every row.

````python
countries = ['France', 'Panama', 'Cyprus']

for country in countries:              # "for each country in the list"
    print("Reviewing:", country)
````

````python
# enumerate gives you the position AND the value
for position, country in enumerate(countries):
    print(position, country)

# Build a result while looping (an audit total)
amounts = [1200, 9950, 450, 9800]
total = 0
for a in amounts:
    if a > 5000:
        total = total + a             # accumulate only large amounts
print("Total of large amounts:", total)
````

````python
# A while loop repeats AS LONG AS a condition holds
counter = 3
while counter > 0:
    print("Countdown:", counter)
    counter = counter - 1             # must move toward stopping, or it runs forever
print("Done")
````

## 9. Functions (`def`)

A **function** is a reusable block of code with a name. You define it with `def`,
give it **inputs** (arguments) and it can **return** a result.

> 💡 **Excel equivalent**: writing your own formula once and reusing it everywhere —
> like a named LAMBDA.

````python
def to_eur(amount, rate):
    """Convert an amount into EUR using the given rate."""
    return round(amount * rate, 2)

# Call the function with different inputs
print(to_eur(1000, 0.92))     # 920.0
print(to_eur(1000, 1.17))     # 1170.0
````

````python
# Default argument values, and a function that returns a category
def risk_category(amount, threshold=10000):
    if amount >= threshold:
        return "High"
    elif amount >= threshold * 0.9:
        return "Watch"
    return "Normal"

print(risk_category(9950))          # uses the default threshold of 10000 -> "Watch"
print(risk_category(500))           # -> "Normal"
print(risk_category(9950, 8000))    # custom threshold -> "High"
````

## 11. Reading an error message

Errors are normal — even for experts. Python prints a **traceback**. Read it **bottom-up**:
the **last line** names the error type and message; the lines above show where it happened.

Common ones:
- `NameError`: a variable/typo that doesn't exist yet
- `KeyError`: a dictionary key (or DataFrame column) that isn't there
- `TypeError`: mixing incompatible types (e.g. text + number)
- `SyntaxError`: a typo in the code structure (missing `:`, `)`, quote…)
- `IndentationError`: misaligned block

````python
# Uncomment ONE line at a time, run it, and read the last line of the error:

# print(montant)              # NameError: name 'montant' is not defined
# rates = {"EUR": 1.0}
# print(rates["USD"])         # KeyError: 'USD'
# print("Amount: " + 100)     # TypeError: can only concatenate str to str

print("Tip: the FIX usually follows directly from the last line of the traceback.")
````

````python
# The safe way to handle an expected error: try / except
rates = {"EUR": 1.0, "GBP": 1.17}
currency = "USD"

try:
    rate = rates[currency]
except KeyError:
    rate = 1.0                       # fallback if the currency is unknown
    print(f"Unknown currency {currency}, using rate 1.0")

print("Rate used:", rate)
````
