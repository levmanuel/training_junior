## Exercise 7 — Loops

```python
amounts = [1200, 9950, 450, 9800, 15000, 300]

total = 0
for a in amounts:
    total += a

count_above = sum(1 for a in amounts if a > 9000)

print("Total:", total)
print("Above 9000:", count_above)
```

## Exercise 8 — Functions

```python
def to_eur(amount, rate):
    return round(amount * rate, 2)

def risk_category(amount, threshold=10000):
    if amount >= threshold:
        return "High"
    elif amount >= threshold * 0.9:
        return "Watch"
    return "Normal"

print(to_eur(1000, 0.92))
print(risk_category(9950))
```

## Exercise 9 — Synthesis

```python
transactions = [
    {"id": 1, "amount": 1200.0, "currency": "USD"},
    {"id": 2, "amount": 9950.0, "currency": "EUR"},
    {"id": 3, "amount": 800.0,  "currency": "GBP"},
    {"id": 4, "amount": 15000.0, "currency": "EUR"},
]
rates = {"EUR": 1.0, "USD": 0.92, "GBP": 1.17}

def to_eur(amount, rate):
    return round(amount * rate, 2)

total_large = 0.0
for tx in transactions:
    eur = to_eur(tx["amount"], rates[tx["currency"]])
    print(f"Tx {tx['id']}: {eur} EUR")
    if eur > 1000:
        total_large += eur

print("Total (> 1000 EUR):", round(total_large, 2))
```
