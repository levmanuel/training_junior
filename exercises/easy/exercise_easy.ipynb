## 0. Getting started and generating the data

Run this cell **first**. It generates the branch dataset.

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
np.random.seed(2024)

n = 180

first_names = ['Marie', 'Jean', 'Sophie', 'Pierre', 'Claire', 'Luc', 'Emma', 'Paul',
           'Isabelle', 'Nicolas', 'Camille', 'Antoine', 'Julie', 'Thomas', 'Laura']
last_names = ['Martin', 'Bernard', 'Dubois', 'Thomas', 'Robert', 'Petit', 'Richard',
        'Simon', 'Laurent', 'Michel', 'Garcia', 'David', 'Moreau', 'Leroy']

dates = pd.to_datetime('2024-03-01') + pd.to_timedelta(
    np.random.randint(0, 62, n), unit='D'
)

operation_types = np.random.choice(
    ['Withdrawal', 'Deposit', 'Check Deposit', 'Advisory', 'Counter Transfer'],
    n, p=[0.30, 0.25, 0.15, 0.20, 0.10]
)

amounts = np.where(
    operation_types == 'Advisory',
    0.0,
    np.round(np.random.lognormal(mean=5.5, sigma=0.9, size=n), 2)
)

client_names = [
    f"{np.random.choice(first_names)} {np.random.choice(last_names)}" for _ in range(n)
]

branch_ops = pd.DataFrame({
    'operation_id':    range(1001, 1001 + n),
    'date':            dates,
    'client_id':       np.random.randint(10000, 10050, n),
    'client_name':     client_names,
    'operation_type':  operation_types,
    'amount':          amounts,
    'advisor':         np.random.choice(['Dupont A.', 'Lefevre B.', 'Morin C.', 'Blanc D.'], n),
    'status':          np.random.choice(['Approved', 'Pending', 'Rejected'], n, p=[0.82, 0.12, 0.06]),
})

# some intentional missing values
branch_ops.loc[np.random.choice(branch_ops.index, 8, replace=False), 'client_name'] = np.nan

branch_ops = branch_ops.sample(frac=1, random_state=7).reset_index(drop=True)

print('Branch dataset ready:', branch_ops.shape[0], 'operations,', branch_ops.shape[1], 'columns')
```
