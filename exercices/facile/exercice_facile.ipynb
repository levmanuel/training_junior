## 0. Mise en route et génération des données

Exécutez cette cellule **en premier**. Elle génère le jeu de données de l'agence.

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
np.random.seed(2024)

n = 180

prenoms = ['Marie', 'Jean', 'Sophie', 'Pierre', 'Claire', 'Luc', 'Emma', 'Paul',
           'Isabelle', 'Nicolas', 'Camille', 'Antoine', 'Julie', 'Thomas', 'Laura']
noms = ['Martin', 'Bernard', 'Dubois', 'Thomas', 'Robert', 'Petit', 'Richard',
        'Simon', 'Laurent', 'Michel', 'Garcia', 'David', 'Moreau', 'Leroy']

dates = pd.to_datetime('2024-03-01') + pd.to_timedelta(
    np.random.randint(0, 62, n), unit='D'
)

types_op = np.random.choice(
    ['Retrait', 'Dépôt', 'Remise chèque', 'Conseil', 'Virement guichet'],
    n, p=[0.30, 0.25, 0.15, 0.20, 0.10]
)

montants = np.where(
    types_op == 'Conseil',
    0.0,
    np.round(np.random.lognormal(mean=5.5, sigma=0.9, size=n), 2)
)

noms_clients = [
    f"{np.random.choice(prenoms)} {np.random.choice(noms)}" for _ in range(n)
]

agence = pd.DataFrame({
    'operation_id':    range(1001, 1001 + n),
    'date':            dates,
    'client_id':       np.random.randint(10000, 10050, n),
    'nom_client':      noms_clients,
    'type_operation':  types_op,
    'montant':         montants,
    'conseiller':      np.random.choice(['Dupont A.', 'Lefevre B.', 'Morin C.', 'Blanc D.'], n),
    'statut':          np.random.choice(['Validé', 'En attente', 'Rejeté'], n, p=[0.82, 0.12, 0.06]),
})

# quelques valeurs manquantes volontaires
agence.loc[np.random.choice(agence.index, 8, replace=False), 'nom_client'] = np.nan

agence = agence.sample(frac=1, random_state=7).reset_index(drop=True)

print('Jeu de données agence prêt :', agence.shape[0], 'opérations,', agence.shape[1], 'colonnes')
```
